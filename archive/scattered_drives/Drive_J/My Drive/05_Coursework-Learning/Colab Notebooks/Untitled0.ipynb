```python
from nltk.tokenize import word_tokenize
text ="Natural Language Processing is very intersting"
tokens = word_tokenize(text)
print(tokens)
```

```python
from nltk.tokenize import sent_tokenize
prargraph="One of the top trending news articles of today covers an intense, fatal confrontation in Gurugram, India, where four armed gangsters were killed and three police officers were injured in a massive shootout following a targeted attack on a local businessman's residence"
sentences = sent_tokenize(prargraph)
print(sentences)
```
