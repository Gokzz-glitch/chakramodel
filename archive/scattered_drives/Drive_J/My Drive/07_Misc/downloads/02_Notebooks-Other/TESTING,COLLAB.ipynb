```python
import os
import shutil
import subprocess

# 1. Ensure Google Drive is mounted
from google.colab import drive
drive.mount('/content/drive')

print("Google Drive mounted! Hunting for the weights...")

# 2. Search for the weights file in your entire Google Drive
target_file = 'chakra_transformer_best.pth'
found_path = None

# We search /content/drive/MyDrive recursively
for root, dirs, files in os.walk('/content/drive/MyDrive'):
    if target_file in files:
        found_path = os.path.join(root, target_file)
        print(f"✅ FOUND IT! The weights are hiding here: {found_path}")
        break

# 3. If we found it, copy it and run the script
if found_path:
    weights_dir = '/content/weights'
    os.makedirs(weights_dir, exist_ok=True)

    # Get the folder where the weights are stored
    found_folder = os.path.dirname(found_path)

    print("Copying weights to the correct Colab folder...")
    # Copy both the Segmenter and YOLO weights
    shutil.copy(os.path.join(found_folder, 'chakra_transformer_best.pth'), os.path.join(weights_dir, 'chakra_transformer_best.pth'))

    yolo_path = os.path.join(found_folder, 'best.pt')
    if os.path.exists(yolo_path):
        shutil.copy(yolo_path, os.path.join(weights_dir, 'best.pt'))
        print("✅ Copied YOLO weights too!")
    else:
        print("⚠️ Warning: Could not find best.pt in that same folder.")

    print("Running the strict verification...")
    # Run the verification
    result = subprocess.run(['python', '/content/src/verify_strict.py'], capture_output=True, text=True)
    print(result.stdout)
    if result.stderr:
        print("ERRORS:\n", result.stderr)

else:
    print("❌ CRITICAL FAILURE: Could not find 'chakra_transformer_best.pth' anywhere in your Google Drive!")
    print("ROOT CAUSE IDENTIFIED: The Google Account you used to mount Drive in Colab does NOT contain the weights.")
    print("SOLUTION: Either you are logged into the wrong Google Account in Colab, or the Google Drive desktop app never actually finished syncing the files.")
```
