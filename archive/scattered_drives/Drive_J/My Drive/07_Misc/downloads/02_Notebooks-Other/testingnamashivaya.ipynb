```python
import os

print("--- Checking Attached Kaggle Datasets ---")
base_path = "/kaggle/input/datasets/gokulrocky/"

if not os.path.exists(base_path):
    print("❌ ERROR: Could not find the base directory. Make sure you attached the datasets from the 'Your Datasets' tab!")
    print("Check /kaggle/input/ manually:")
    print(os.listdir("/kaggle/input/"))
else:
    print("✅ Base directory found! Here is the architecture of what is attached:\n")
    for root, dirs, files in os.walk(base_path):
        # Calculate indentation for a clean tree view
        level = root.replace(base_path, '').count(os.sep)
        indent = ' ' * 4 * level
        print(f"{indent}📂 {os.path.basename(root)}/")
        
        if files:
            sub_indent = ' ' * 4 * (level + 1)
            file_count = len(files)
            # Show a few files as proof, but don't flood the screen if there are thousands of images
            if file_count > 4:
                for f in files[:2]:
                    print(f"{sub_indent}📄 {f}")
                print(f"{sub_indent}... and {file_count - 2} more files.")
            else:
                for f in files:
                    print(f"{sub_indent}📄 {f}")
```
