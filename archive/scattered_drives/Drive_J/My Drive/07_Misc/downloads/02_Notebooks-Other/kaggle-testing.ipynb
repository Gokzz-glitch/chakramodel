```python
from pathlib import Path
from collections import Counter
import cv2

ROOT = Path("/kaggle/input")

image_exts = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"}
video_exts = {".avi", ".mp4", ".mov", ".mkv"}
mask_words = {"mask", "masks", "groundtruth", "ground_truth", "segmentation", "label", "labels"}

for dataset in sorted(p for p in ROOT.iterdir() if p.is_dir()):
    files = [p for p in dataset.rglob("*") if p.is_file()]
    images = [p for p in files if p.suffix.lower() in image_exts]
    videos = [p for p in files if p.suffix.lower() in video_exts]
    mask_like = [p for p in images if any(w in p.as_posix().lower() for w in mask_words)]

    sizes = Counter()
    for path in images[:500]:
        image = cv2.imread(str(path))
        if image is not None:
            sizes[tuple(image.shape[:2])] += 1

    print(f"\nDATASET: {dataset.name}")
    print(f"files={len(files)}, images={len(images)}, videos={len(videos)}, mask_like={len(mask_like)}")
    print(f"top resolutions={sizes.most_common(5)}")
    print("sample folders:")
    for folder in sorted({str(p.parent.relative_to(dataset)) for p in files})[:15]:
        print(" ", folder)
```
