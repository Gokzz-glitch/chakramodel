```python
"""
KAGGLE FINAL WORKING V2 - UNet Testing
Fixed: Match images & masks by number (not filename)
"""

!pip install -q tensorflow==2.20.0 keras albumentations scikit-image matplotlib opencv-python pandas h5py -U

import os
import re
import numpy as np
import tensorflow as tf
from sklearn.metrics import precision_score, recall_score, f1_score
import matplotlib.pyplot as plt
import cv2
import warnings
import pandas as pd
import h5py
warnings.filterwarnings('ignore')

print(f"GPU: {tf.config.list_physical_devices('GPU')}")
print(f"TensorFlow: {tf.__version__}\n")

# ============================================================================
# FIND IMAGES & MASKS
# ============================================================================
print("="*60)
print("FINDING IMAGES & MASKS")
print("="*60)

# Find images
IMAGE_DIR = None
img_base = "/kaggle/input/datasets/gokulrocky/kvasir-seg-images-zip"

for root, dirs, files in os.walk(img_base):
    jpg_files = [f for f in files if f.endswith('.jpg')]
    if jpg_files:
        IMAGE_DIR = root
        print(f"✓ Found {len(jpg_files)} images in: {IMAGE_DIR}")
        break

if not IMAGE_DIR:
    print("✗ Could not find images!")
    import sys
    sys.exit()

# Find masks
MASK_DIR = "/kaggle/input/datasets/gokulrocky/kvasir-seg-masks/masks"
all_mask_files = sorted(os.listdir(MASK_DIR))
mask_files = [f for f in all_mask_files if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
print(f"✓ Found {len(mask_files)} masks in: {MASK_DIR}")

# Extract numbers from filenames
def extract_number(filename):
    match = re.search(r'(\d+)', filename)
    return match.group(1) if match else None

# Create mapping: number -> filename
image_map = {}  # number -> image filename
mask_map = {}   # number -> mask filename

all_images = sorted([f for f in os.listdir(IMAGE_DIR) if f.endswith('.jpg')])
for img in all_images:
    num = extract_number(img)
    if num:
        image_map[num] = img

for mask in mask_files:
    num = extract_number(mask)
    if num:
        mask_map[num] = mask

# Find matching pairs (by number)
matching_numbers = set(image_map.keys()) & set(mask_map.keys())
print(f"✓ Image basenames: {all_images[:3]}")
print(f"✓ Mask basenames: {mask_files[:3]}")
print(f"✓ Matching image-mask pairs (by number): {len(matching_numbers)}")

# ============================================================================
# BUILD MODEL WITH CONV2DTRANSPOSE FIX
# ============================================================================
print("\n" + "="*60)
print("LOADING MODEL")
print("="*60)

from tensorflow.keras.layers import Conv2DTranspose

class CustomConv2DTranspose(Conv2DTranspose):
    def __init__(self, *args, **kwargs):
        kwargs.pop('groups', None)
        super().__init__(*args, **kwargs)

WEIGHTS_PATH = "/kaggle/input/datasets/gokulrocky/unet-weight/unet (1).h5"

try:
    custom_objects = {'Conv2DTranspose': CustomConv2DTranspose}
    model = tf.keras.models.load_model(WEIGHTS_PATH, custom_objects=custom_objects)
    print("✓ Model loaded with custom objects")
except Exception as e:
    print(f"✗ Error: {str(e)[:100]}")
    import sys
    sys.exit()

print("\nModel structure:")
model.summary()

# ============================================================================
# PREPARE TEST DATA (70/15/15 split)
# ============================================================================
print("\n" + "="*60)
print("PREPARING TEST DATA")
print("="*60)

# Use only matching pairs
matching_list = sorted(list(matching_numbers))
print(f"Total paired images: {len(matching_list)}")

# Create 70/15/15 split
np.random.seed(42)
indices = np.arange(len(matching_list))
np.random.shuffle(indices)

split1 = int(0.70 * len(indices))
split2 = int(0.85 * len(indices))

test_idx = indices[split2:]
test_numbers = [matching_list[i] for i in test_idx]

print(f"Test samples: {len(test_numbers)}")

# ============================================================================
# LOAD TEST DATA - CORRECTED SIZE (256x256)
# ============================================================================
print("\nLoading test images and masks...")

IMG_HEIGHT, IMG_WIDTH = 256, 256
X_test = []
y_test = []

mean = np.array([0.485, 0.456, 0.406])
std = np.array([0.229, 0.224, 0.225])

loaded_count = 0
failed_count = 0

for num in test_numbers:
    try:
        # Get filenames
        img_filename = image_map[num]
        mask_filename = mask_map[num]

        # Load image
        img_path = os.path.join(IMAGE_DIR, img_filename)
        img = cv2.imread(img_path)
        if img is None:
            failed_count += 1
            continue

        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = cv2.resize(img, (IMG_WIDTH, IMG_HEIGHT))
        img = img.astype('float32') / 255.0
        img = (img - mean) / std
        X_test.append(img)

        # Load mask
        mask_path = os.path.join(MASK_DIR, mask_filename)
        mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)
        if mask is None:
            failed_count += 1
            X_test.pop()  # Remove image since mask failed
            continue

        mask = cv2.resize(mask, (IMG_WIDTH, IMG_HEIGHT))
        mask = mask.astype('float32') / 255.0
        mask = (mask > 0.5).astype('float32')
        y_test.append(mask)
        loaded_count += 1

    except Exception as e:
        failed_count += 1
        continue

X_test = np.array(X_test)
y_test = np.array(y_test)

print(f"✓ Successfully loaded: {loaded_count} image-mask pairs")
print(f"✗ Failed: {failed_count}")
print(f"  X_test: {X_test.shape}")
print(f"  y_test: {y_test.shape}")

if len(X_test) == 0:
    print("✗ No data loaded!")
    import sys
    sys.exit()

# ============================================================================
# INFERENCE
# ============================================================================
print("\n" + "="*60)
print("RUNNING INFERENCE")
print("="*60)

y_pred = model.predict(X_test, batch_size=4, verbose=1)
y_pred = (y_pred > 0.5).astype('float32')

# ============================================================================
# METRICS
# ============================================================================
def dice_coefficient(y_true, y_pred):
    smooth = 1.0
    intersection = np.sum(y_true * y_pred)
    return (2.0 * intersection + smooth) / (np.sum(y_true) + np.sum(y_pred) + smooth)

def iou_score(y_true, y_pred):
    smooth = 1.0
    intersection = np.sum(y_true * y_pred)
    union = np.sum(y_true) + np.sum(y_pred) - intersection
    return (intersection + smooth) / (union + smooth)

print("\n" + "="*60)
print("RESULTS")
print("="*60)

y_pred = np.squeeze(y_pred, axis=-1)

y_test_flat = y_test.flatten()
y_pred_flat = y_pred.flatten()

dice = dice_coefficient(y_test, y_pred)
iou = iou_score(y_test, y_pred)
precision = precision_score(y_test_flat, y_pred_flat, zero_division=0)
recall = recall_score(y_test_flat, y_pred_flat, zero_division=0)
f1 = f1_score(y_test_flat, y_pred_flat, zero_division=0)

print(f"\nDice Coefficient:  {dice:.4f}")
print(f"IoU Score:         {iou:.4f}")
print(f"Precision:         {precision:.4f}")
print(f"Recall:            {recall:.4f}")
print(f"F1 Score:          {f1:.4f}")

# Save results
results_df = pd.DataFrame({
    'Metric': ['Dice', 'IoU', 'Precision', 'Recall', 'F1'],
    'Score': [dice, iou, precision, recall, f1]
})
results_df.to_csv('test_results.csv', index=False)
print("\n✓ Saved: test_results.csv")

# Per-sample analysis
per_sample_dice = [dice_coefficient(y_test[i:i+1], y_pred[i:i+1]) for i in range(len(X_test))]
per_sample_iou = [iou_score(y_test[i:i+1], y_pred[i:i+1]) for i in range(len(X_test))]

print(f"\nPer-Sample Statistics:")
print(f"Dice  - Min: {np.min(per_sample_dice):.4f}, Max: {np.max(per_sample_dice):.4f}, Mean: {np.mean(per_sample_dice):.4f}, Std: {np.std(per_sample_dice):.4f}")
print(f"IoU   - Min: {np.min(per_sample_iou):.4f}, Max: {np.max(per_sample_iou):.4f}, Mean: {np.mean(per_sample_iou):.4f}, Std: {np.std(per_sample_iou):.4f}")

# Visualize predictions
num_samples = min(6, len(X_test))
fig, axes = plt.subplots(num_samples, 3, figsize=(15, 4*num_samples))

for i in range(num_samples):
    img_display = X_test[i] * std+ mean
    img_display = np.clip(img_display, 0, 1)

    axes[i, 0].imshow(img_display)
    axes[i, 0].set_title("Original")
    axes[i, 0].axis('off')

    axes[i, 1].imshow(y_test[i], cmap='gray')
    axes[i, 1].set_title("Ground Truth")
    axes[i, 1].axis('off')

    axes[i, 2].imshow(y_pred[i], cmap='gray')
    axes[i, 2].set_title("Prediction")
    axes[i, 2].axis('off')

plt.tight_layout()
plt.savefig('predictions.png', dpi=100, bbox_inches='tight')
plt.show()
print("✓ Saved: predictions.png")

# Visualize metrics distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(per_sample_dice, bins=20, color='skyblue', edgecolor='black', alpha=0.7)
axes[0].axvline(np.mean(per_sample_dice), color='red', linestyle='--', linewidth=2, label=f'Mean: {np.mean(per_sample_dice):.4f}')
axes[0].set_xlabel('Dice Score')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Dice Distribution')
axes[0].legend()
axes[0].grid(alpha=0.3)

axes[1].hist(per_sample_iou, bins=20, color='lightcoral', edgecolor='black', alpha=0.7)
axes[1].axvline(np.mean(per_sample_iou), color='red', linestyle='--', linewidth=2, label=f'Mean: {np.mean(per_sample_iou):.4f}')
axes[1].set_xlabel('IoU Score')
axes[1].set_ylabel('Frequency')
axes[1].set_title('IoU Distribution')
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig('metrics_distribution.png', dpi=100, bbox_inches='tight')
plt.show()
print("✓ Saved: metrics_distribution.png")

print("\n✓✓✓ COMPLETE ✓✓✓")
```
