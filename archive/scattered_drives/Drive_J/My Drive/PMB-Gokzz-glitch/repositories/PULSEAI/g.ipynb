```python
import time, math, itertools

counter = 0
spin = itertools.cycle(['⠋','⠙','⠹','⠸','⠼','⠴','⠦','⠧','⠇','⠏'])

print("Keep-alive running. Press ■ (Stop) to end.\n")

while True:
    # Light CPU work — won't overheat or hog resources
    _ = sum(math.sin(i * 0.001) for i in range(500))

    counter += 1
    print(f"\r  {next(spin)}  alive for {counter * 30}s", end='', flush=True)

    time.sleep(30)  # checks in every 30 seconds
```
