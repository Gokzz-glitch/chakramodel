```python
from google.colab import drive
import os

# Mount Google Drive
drive.mount('/content/drive')

# --- IMPORTANT: Update this path to where your dataset is in Google Drive ---
dataset_path = '/content/drive/MyDrive/polyp_dataset_for_evaluation'

def clean_filenames(root_dir):
    # Walk bottom-up to ensure parent directories are renamed AFTER their contents
    for dirpath, dirnames, filenames in os.walk(root_dir, topdown=False):

        # 1. Rename files
        for filename in filenames:
            if '\\' in filename:
                new_name = filename.replace('\\', '_')
                old_file_path = os.path.join(dirpath, filename)
                new_file_path = os.path.join(dirpath, new_name)
                os.rename(old_file_path, new_file_path)
                print(f"Renamed file: '{filename}' -> '{new_name}'")

        # 2. Rename directories
        for dirname in dirnames:
            if '\\' in dirname:
                new_name = dirname.replace('\\', '_')
                old_dir_path = os.path.join(dirpath, dirname)
                new_dir_path = os.path.join(dirpath, new_name)
                os.rename(old_dir_path, new_dir_path)
                print(f"Renamed directory: '{dirname}' -> '{new_name}'")

# Execute the renaming
clean_filenames(dataset_path)
print("All files and directories checked and renamed successfully!")
```
