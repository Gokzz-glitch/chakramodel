```python
import numpy as np
import pandas as pd

# Set a random seed for reproducibility
np.random.seed(42)

# Number of samples per class
samples_per_class = 1667

# ---------------------------------------------------------
# Class 0: Normal State
# pH: ~6.8 to 7.2 | TDS: ~1000 to 1500 ppm
# ---------------------------------------------------------
normal_ph = np.random.normal(loc=7.0, scale=0.2, size=samples_per_class)
normal_tds = np.random.normal(loc=1250, scale=150, size=samples_per_class)
normal_temp = np.random.normal(loc=36.0, scale=0.5, size=samples_per_class) # Fresh sample temp
normal_labels = np.zeros(samples_per_class, dtype=int)

# ---------------------------------------------------------
# Class 1: Dehydrated State
# Lower saliva volume means higher ionic concentration (TDS spikes).
# pH drops slightly.
# ---------------------------------------------------------
dehydrated_ph = np.random.normal(loc=6.6, scale=0.2, size=samples_per_class)
dehydrated_tds = np.random.normal(loc=2200, scale=200, size=samples_per_class)
dehydrated_temp = np.random.normal(loc=36.0, scale=0.5, size=samples_per_class)
dehydrated_labels = np.ones(samples_per_class, dtype=int)

# ---------------------------------------------------------
# Class 2: Metabolic Imbalance (Acidosis Proxy)
# pH drops significantly below 6.2. TDS remains variable.
# ---------------------------------------------------------
acidosis_ph = np.random.normal(loc=5.8, scale=0.3, size=samples_per_class)
acidosis_tds = np.random.normal(loc=1400, scale=250, size=samples_per_class)
acidosis_temp = np.random.normal(loc=36.0, scale=0.5, size=samples_per_class)
acidosis_labels = np.full(samples_per_class, 2, dtype=int)

# ---------------------------------------------------------
# Combine and format the dataset
# ---------------------------------------------------------
ph_data = np.concatenate([normal_ph, dehydrated_ph, acidosis_ph])
tds_data = np.concatenate([normal_tds, dehydrated_tds, acidosis_tds])
temp_data = np.concatenate([normal_temp, dehydrated_temp, acidosis_temp])
labels = np.concatenate([normal_labels, dehydrated_labels, acidosis_labels])

# Create a Pandas DataFrame
df = pd.DataFrame({
    'pH_Value': np.round(ph_data, 2),
    'TDS_ppm': np.round(tds_data, 0),
    'Temp_C': np.round(temp_data, 1),
    'Target_Class': labels
})

# Shuffle the dataset so the classes aren't in perfectly sequential blocks
df = df.sample(frac=1).reset_index(drop=True)

# Save to CSV for the neural network pipeline
df.to_csv('synthetic_saliva_dataset.csv', index=False)

print("Dataset successfully generated and saved as 'synthetic_saliva_dataset.csv'!")
print(df.head(10))
print("\nClass Distribution:")
print(df['Target_Class'].value_counts())
```
