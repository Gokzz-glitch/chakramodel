```python
!pip install -q playwright
!playwright install chromium

from playwright.sync_api import sync_playwright
import time

def capture_drive_image(drive_file_url, output_path):
    with sync_playwright() as p:
        browser = p.chromium.launch(headless=True)
        # Set viewport to match target scan aspect ratio high res
        page = browser.new_page(viewport={"width": 1920, "height": 1080})

        page.goto(drive_file_url)
        time.sleep(3)  # Wait for full dynamic range scan to load

        # Capture strictly the rendered image element, bypassing UI
        image_element = page.query_selector('img')
        if image_element:
            image_element.screenshot(path=output_path)
            print(f"Successfully saved clean image to {output_path}")
        else:
            page.screenshot(path=output_path)
        browser.close()

# Example usage for a restricted scan link:
# capture_drive_image("https://drive.google.com/file/d/YOUR_FILE_ID/view", "fchd_sample.jpg")
```
