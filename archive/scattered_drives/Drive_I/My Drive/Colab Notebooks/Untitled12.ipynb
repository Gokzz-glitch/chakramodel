```python
class solution:
  def len_of_substring(self, s: str) -> int:
    n=len(s)
    left=0
    maxlen=0
    charset=set()

    for i in range(n):
      if s[i] not in charset:
        charset.add(s[i])
        maxlen=max(maxlen,i-left+1)
      else:
        while s[i] in charset:
          charset.remove(s[left])
          left+=1
        charset.add(s[i])
    return maxlen
```

```python
len_of_substring("abcabcabb)
```
