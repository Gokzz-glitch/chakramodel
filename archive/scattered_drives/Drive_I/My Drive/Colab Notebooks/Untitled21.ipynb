```python
import os
from autodistill_grounding_dino import GroundingDINO
from autodistill.detection import CaptionOntology
from autodistill_yolov8 import YOLOv8

# 1. Map natural language text prompts to your dataset class names
# Tip: Adding descriptive adjectives helps Grounding DINO identify low-contrast structures
ontology = CaptionOntology({
    "fetal thorax chest contour": "thorax",
    "left ventricle of fetal heart": "left_ventricle",
    "right ventricle of fetal heart": "right_ventricle",
    "left atrium of fetal heart": "left_atrium",
    "right atrium of fetal heart": "right_atrium",
    "fetal aorta vessel": "aorta",
    "fetal pulmonary artery": "pulmonary_artery",
    "superior vena cava vessel": "svc",
    "ventricular septal defect cardiac anomaly": "potential_chd"
})

# 2. Initialize Grounding DINO as the Base (Teacher) Model
base_model = GroundingDINO(
    ontology=ontology,
    box_threshold=0.35,  # Confidence threshold for bounding boxes
    text_threshold=0.25  # Text match threshold
)

# 3. Auto-label your unannotated drive folder images
# This will output images and YOLO-formatted .txt label files
dataset_dir = "/content/drive/MyDrive/dataset/02_processed/train"

print("🚀 Starting auto-annotation process...")
dataset = base_model.label(
    input_folder="/content/drive/MyDrive/dataset/01_raw",  # Path to your raw image folder
    output_folder=dataset_dir
)

print(f"✅ Auto-annotation complete! Dataset saved at: {dataset_dir}")
```

```python
# 4. Target (Student) Model Training
target_model = YOLOv8("yolov8m.pt")

# Train YOLOv8 on the auto-labeled dataset
target_model.train(
    dataset_dir=f"{dataset_dir}/data.yaml",
    epochs=50,
    imgsz=640
)
```
