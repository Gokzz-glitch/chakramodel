```python
# === PURE CHAKRA-TRANSFORMER EVALUATION ===
# Load ViT-Large ChakraTransformer, evaluate on Kvasir-SEG last 10% + CVC-ClinicDB
import sys, os
sys.path.insert(0, f"{WP}/src")
sys.path.insert(0, f"{WP}/src/chakra_transformer")

# Import model
from chakra_transformer.transformer_segmenter import ChakraTransformerSegmenter

# Build with pretrained=False (we'll load our own weights)
model = ChakraTransformerSegmenter(backbone_name='vit_large_patch16_384',
                                   pretrained=False, num_classes=1)
# Load checkpoint - strip 'module.' prefix from DataParallel
sd_raw = torch.load(f"{WP}/weights/chakra_transformer_best.pth", map_location='cpu', weights_only=False)
sd = {k.replace('module.', '', 1): v for k, v in sd_raw.items()}
missing, unexpected = model.load_state_dict(sd, strict=False)
print(f"Missing keys: {len(missing)}, Unexpected keys: {len(unexpected)}")
if missing[:3]: print(f"  first missing: {missing[:3]}")
if unexpected[:3]: print(f"  first unexpected: {unexpected[:3]}")

model = model.cuda().eval()
n_params = sum(p.numel() for p in model.parameters())
print(f"\nModel: ChakraTransformerSegmenter (ViT-L/16 384)")
print(f"Parameters: {n_params/1e6:.2f}M")
print(f"Device: cuda")
```

```python
# === Evaluation metrics + preprocessing ===
import numpy as np
import cv2
from PIL import Image
from pathlib import Path

IMG_SIZE = 384
IMAGENET_MEAN = np.array([0.485, 0.456, 0.406], dtype=np.float32)
IMAGENET_STD  = np.array([0.229, 0.224, 0.225], dtype=np.float32)

def preprocess(img_bgr):
    """BGR uint8 HxWx3 -> tensor 1x3x384x384 (ImageNet normalized)."""
    img = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (IMG_SIZE, IMG_SIZE), interpolation=cv2.INTER_LINEAR)
    img = img.astype(np.float32) / 255.0
    img = (img - IMAGENET_MEAN) / IMAGENET_STD
    return torch.from_numpy(img.transpose(2,0,1)).unsqueeze(0).cuda()

def dice(pred, gt, eps=1e-7):
    p, g = pred.astype(bool), gt.astype(bool)
    inter = (p & g).sum()
    return (2 * inter + eps) / (p.sum() + g.sum() + eps)

def iou(pred, gt, eps=1e-7):
    p, g = pred.astype(bool), gt.astype(bool)
    inter = (p & g).sum(); union = (p | g).sum()
    return (inter + eps) / (union + eps)

def f_beta(pred, gt, beta2=0.3, eps=1e-7):
    p, g = pred.astype(bool), gt.astype(bool)
    tp = (p & g).sum(); fp = (p & ~g).sum(); fn = (~p & g).sum()
    prec = tp / (tp + fp + eps); rec = tp / (tp + fn + eps)
    return ((1+beta2) * prec * rec) / (beta2 * prec + rec + eps)

print("Metrics defined. Testing on dummy image...")
dummy = np.zeros((512, 512, 3), dtype=np.uint8)
t = preprocess(dummy)
print(f"Preprocessed shape: {t.shape}, device: {t.device}")
with torch.no_grad():
    out = model(t)
print(f"Model output shape: {out.shape}")
print(f"Output range: [{out.min().item():.3f}, {out.max().item():.3f}]")
```

```python
# === RUN FULL EVALUATION ===
from tqdm.auto import tqdm

def eval_dataset(name, images_dir, masks_dir, indices=None):
    imgs = sorted([p for p in Path(images_dir).iterdir()
                   if p.suffix.lower() in ('.png','.jpg','.jpeg','.tif','.bmp')])
    if indices is not None:
        imgs = [imgs[i] for i in indices]
    dices, ious, fbs = [], [], []
    for img_path in tqdm(imgs, desc=name):
        img = cv2.imread(str(img_path))
        if img is None: continue
        # Match mask by stem
        mask = None
        for ext in ('.png','.jpg','.jpeg','.tif','.bmp'):
            mp = Path(masks_dir) / (img_path.stem + ext)
            if mp.exists():
                mask = cv2.imread(str(mp), cv2.IMREAD_GRAYSCALE)
                break
        if mask is None: continue
        H, W = mask.shape
        with torch.no_grad():
            t = preprocess(img)
            logits = model(t)
            prob = torch.sigmoid(logits)[0,0].cpu().numpy()
        # Resize to GT size
        prob = cv2.resize(prob, (W, H), interpolation=cv2.INTER_LINEAR)
        pred = (prob > 0.5).astype(np.uint8)
        gt = (mask > 127).astype(np.uint8)
        dices.append(dice(pred, gt))
        ious.append(iou(pred, gt))
        fbs.append(f_beta(pred, gt))
    return np.mean(dices), np.mean(ious), np.mean(fbs), len(dices)

# Kvasir-SEG: last 10% as test (standard PraNet-style held-out)
kv_img = f"{DP}/kvasir-seg/images"
kv_msk = f"{DP}/kvasir-seg/masks"
n_kv = len(sorted(os.listdir(kv_img)))
kv_test_idx = list(range(n_kv - int(0.1*n_kv), n_kv))
print(f"Kvasir-SEG: evaluating last {len(kv_test_idx)} of {n_kv} images (held-out)")

d, i, f, n = eval_dataset("Kvasir-SEG (last 10%)", kv_img, kv_msk, kv_test_idx)
print(f"\n>>> Kvasir-SEG (n={n}): Dice={d:.4f} IoU={i:.4f} Fbeta={f:.4f}")

# CVC-ClinicDB: full 495 (no held-out split known)
cc_img = f"{DP}/cvc-clinicdb/images"
cc_msk = f"{DP}/cvc-clinicdb/masks"
d2, i2, f2, n2 = eval_dataset("CVC-ClinicDB (all)", cc_img, cc_msk)
print(f"\n>>> CVC-ClinicDB (n={n2}): Dice={d2:.4f} IoU={i2:.4f} Fbeta={f2:.4f}")

print("\n============ RESULTS ============")
print(f"{'Dataset':<25} {'N':>5} {'Dice':>8} {'IoU':>8} {'Fbeta':>8}")
print(f"{'Kvasir-SEG (test 10%)':<25} {n:>5} {d:>8.4f} {i:>8.4f} {f:>8.4f}")
print(f"{'CVC-ClinicDB (all)':<25} {n2:>5} {d2:>8.4f} {i2:>8.4f} {f2:>8.4f}")
```

```python
# === MAX ACCURACY: 4-way TTA (horizontal + vertical flips) + MAX FPS: FP16 batch inference ===
# Use both T4 GPUs via nn.DataParallel wrapping

model_dp = torch.nn.DataParallel(model, device_ids=[0, 1])

@torch.no_grad()
def infer_tta_amp(img_bgr):
    """4-way TTA: identity, hflip, vflip, both. FP16. Returns numpy prob map at input size."""
    img = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (IMG_SIZE, IMG_SIZE), interpolation=cv2.INTER_LINEAR)
    img = img.astype(np.float32) / 255.0
    img = (img - IMAGENET_MEAN) / IMAGENET_STD
    x = torch.from_numpy(img.transpose(2,0,1)).unsqueeze(0).cuda()
    # 4 versions batched at once (uses both GPUs via DP)
    xs = torch.cat([x, torch.flip(x,[3]), torch.flip(x,[2]), torch.flip(x,[2,3])], dim=0)
    with torch.cuda.amp.autocast(dtype=torch.float16):
        ys = model_dp(xs)
    ys = torch.sigmoid(ys.float())
    # Reverse the flips
    ys = torch.stack([ys[0], torch.flip(ys[1],[2]), torch.flip(ys[2],[1]), torch.flip(ys[3],[1,2])])
    prob = ys.mean(0)[0].cpu().numpy()
    return prob

def eval_tta(name, images_dir, masks_dir, indices=None):
    imgs = sorted([p for p in Path(images_dir).iterdir()
                   if p.suffix.lower() in ('.png','.jpg','.jpeg','.tif','.bmp')])
    if indices is not None:
        imgs = [imgs[i] for i in indices]
    dices, ious, fbs = [], [], []
    for img_path in tqdm(imgs, desc=name):
        img = cv2.imread(str(img_path))
        if img is None: continue
        mask = None
        for ext in ('.png','.jpg','.jpeg','.tif','.bmp'):
            mp = Path(masks_dir) / (img_path.stem + ext)
            if mp.exists():
                mask = cv2.imread(str(mp), cv2.IMREAD_GRAYSCALE)
                break
        if mask is None: continue
        H, W = mask.shape
        prob = infer_tta_amp(img)
        prob = cv2.resize(prob, (W, H), interpolation=cv2.INTER_LINEAR)
        pred = (prob > 0.5).astype(np.uint8)
        gt = (mask > 127).astype(np.uint8)
        dices.append(dice(pred, gt))
        ious.append(iou(pred, gt))
        fbs.append(f_beta(pred, gt))
    return np.mean(dices), np.mean(ious), np.mean(fbs), len(dices)

print("Running TTA+FP16 evaluation on Kvasir-SEG test 10%...")
d_kv, i_kv, f_kv, n_kv_t = eval_tta("Kvasir-SEG TTA", kv_img, kv_msk, kv_test_idx)
print(f">>> Kvasir-SEG (n={n_kv_t}): Dice={d_kv:.4f} IoU={i_kv:.4f} Fbeta={f_kv:.4f}")

print("\nRunning TTA+FP16 evaluation on CVC-ClinicDB...")
d_cc, i_cc, f_cc, n_cc_t = eval_tta("CVC-ClinicDB TTA", cc_img, cc_msk)
print(f">>> CVC-ClinicDB (n={n_cc_t}): Dice={d_cc:.4f} IoU={i_cc:.4f} Fbeta={f_cc:.4f}")

print("\n============ TTA+FP16 RESULTS ============")
print(f"{'Dataset':<25} {'N':>5} {'Dice':>8} {'IoU':>8} {'Fbeta':>8}")
print(f"{'Kvasir-SEG (TTA)':<25} {n_kv_t:>5} {d_kv:>8.4f} {i_kv:>8.4f} {f_kv:>8.4f}")
print(f"{'CVC-ClinicDB (TTA)':<25} {n_cc_t:>5} {d_cc:>8.4f} {i_cc:>8.4f} {f_cc:>8.4f}")
```

```python
# === MAX FPS BENCHMARK ===
# Test different batch sizes, FP32 vs FP16, single vs dual GPU
import time

def bench_fps(model_ref, batch_size, n_iter=50, use_amp=False, warmup=10):
    x = torch.randn(batch_size, 3, IMG_SIZE, IMG_SIZE, device='cuda')
    # Warmup
    with torch.no_grad():
        for _ in range(warmup):
            if use_amp:
                with torch.amp.autocast('cuda', dtype=torch.float16):
                    _ = model_ref(x)
            else:
                _ = model_ref(x)
    torch.cuda.synchronize()
    # Timed
    t0 = time.perf_counter()
    with torch.no_grad():
        for _ in range(n_iter):
            if use_amp:
                with torch.amp.autocast('cuda', dtype=torch.float16):
                    _ = model_ref(x)
            else:
                _ = model_ref(x)
    torch.cuda.synchronize()
    dt = time.perf_counter() - t0
    fps = n_iter * batch_size / dt
    ms_per = dt / n_iter * 1000
    return fps, ms_per

print("=" * 60)
print("FPS BENCHMARK — ChakraTransformer ViT-L/16 @ 384x384")
print("=" * 60)
configs = [
    ("Single T4, FP32, bs=1",  model,    1,  False),
    ("Single T4, FP16, bs=1",  model,    1,  True),
    ("Single T4, FP16, bs=4",  model,    4,  True),
    ("Single T4, FP16, bs=8",  model,    8,  True),
    ("Dual T4,   FP16, bs=2",  model_dp, 2,  True),
    ("Dual T4,   FP16, bs=8",  model_dp, 8,  True),
    ("Dual T4,   FP16, bs=16", model_dp, 16, True),
]
results = []
for name, m, bs, amp in configs:
    try:
        fps, ms = bench_fps(m, bs, n_iter=30, use_amp=amp)
        results.append((name, bs, fps, ms))
        print(f"{name:<28} bs={bs:>2}  FPS={fps:>7.2f}  {ms:>6.2f} ms/batch")
    except Exception as e:
        print(f"{name:<28} FAILED: {type(e).__name__}: {str(e)[:60]}")
        torch.cuda.empty_cache()

print()
best = max(results, key=lambda r: r[2])
print(f"BEST: {best[0]}  ->  {best[2]:.2f} FPS")
```
